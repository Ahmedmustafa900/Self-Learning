```python
def sales_performance_2025(df):
    """
    This metric compares the total ACV of closed-won deals in the first half
    versus the second half of 2025. It shows whether the sales team improved
    or declined in its ability to generate revenue over time.
    """
    h1_acv = 0
    h2_acv = 0

    for index, row in df.iterrows():
        if row['Status'] == 'Won' and row['CloseDate'].year == 2025:
            if row['CloseDate'].month <= 6:
                h1_acv += row['ACV']
            else:
                h2_acv += row['ACV']

    return h1_acv, h2_acv


# Print the metric results
h1, h2 = sales_performance_2025(df)
print(h1, h2)
```

In this case, the lower ACV in the second half suggests a decline in revenue generation, which would warrant further investigation into win rates, deal sizes, or sales cycle length.”

```python
def monthly_won_acv_2025(dataframe):
    """
    This metric calculates the total ACV closed per month in 2025.
    It helps identify sales performance trends, seasonality, and
    momentum changes throughout the year.
    """
    monthly_acv = {}

    for index, row in dataframe.iterrows():
        if row['Status'] == 'Won' and row['CloseDate'].year == 2025:
            month = row['CloseDate'].month
            monthly_acv[month] = monthly_acv.get(month, 0) + row['ACV']

    return monthly_acv


# Print monthly ACV
print(monthly_won_acv_2025(df))
```

```python
def average_deal_size_2025(dataframe):
    """
    This metric calculates the average ACV of closed-won deals in 2025.
    It shows whether deal quality improved or declined over time.
    """
    total_acv = 0
    deal_count = 0

    for index, row in dataframe.iterrows():
        if row['Status'] == 'Won' and row['CloseDate'].year == 2025:
            total_acv += row['ACV']
            deal_count += 1

    if deal_count == 0:
        return 0

    return total_acv / deal_count


print(average_deal_size_2025(df))
```

```python
def win_rate_2025(dataframe):
    """
    This metric calculates the sales win rate for opportunities
    closed in 2025. It reflects sales effectiveness and deal quality.
    """
    won = 0
    lost = 0

    for index, row in dataframe.iterrows():
        if row['Status'] == 'Won' and row['CloseDate'].year == 2025:
            won += 1
        if row['Status'] == 'Lost' and row['CloseDate'].year == 2025:
            lost += 1

    if won + lost == 0:
        return 0

    return won / (won + lost)


print(win_rate_2025(df))
```
