```python
# GenAI-based document deviation using Hugging Face embeddings

from sentence_transformers import SentenceTransformer
import numpy as np


# Load a Hugging Face sentence embedding model 
_model_name = "sentence-transformers/all-MiniLM-L6-v2"
_hf_model = SentenceTransformer(_model_name)


def _embed_text_hf(text: str) -> np.ndarray:
    """Get a semantic embedding vector for the given text using a Hugging Face model."""
    # SentenceTransformer returns a NumPy array directly when convert_to_tensor=False
    emb = _hf_model.encode(text, convert_to_tensor=False, normalize_embeddings=False)
    return np.array(emb, dtype=float)


def _cosine_similarity_vec(v1: np.ndarray, v2: np.ndarray) -> float:
    """Cosine similarity for two embedding vectors."""
    if v1.size == 0 or v2.size == 0:
        return 0.0
    denom = (np.linalg.norm(v1) * np.linalg.norm(v2))
    if denom == 0:
        return 0.0
    return float(np.dot(v1, v2) / denom)


def compute_document_deviation_genai(doc1: str, doc2: str) -> dict:
    """Compute semantic similarity and deviation between two documents using a Hugging Face model.

    dict with keys:
      - "similarity": float in [0, 1] (higher = more similar)
      - "deviation": float in [0, 1] (higher = more different)
    """
    emb1 = _embed_text_hf(doc1)
    emb2 = _embed_text_hf(doc2)

    similarity = _cosine_similarity_vec(emb1, emb2)
    deviation = 1.0 - similarity

    return {
        "similarity": similarity,
        "deviation": deviation,
    }

if __name__ == "__main__":
    doc_a = "This is the first sample document for comparison."
    doc_b = "This is another document used to compare the deviation between texts."

    result_genai = compute_document_deviation_genai(doc_a, doc_b)
    print("[HF GenAI] Similarity:", result_genai["similarity"])
    print("[HF GenAI] Deviation:", result_genai["deviation"])
```
